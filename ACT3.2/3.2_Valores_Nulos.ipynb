**Métodos de Sustitución de Valores Nulos**

Aplicación de métodos para reemplazar valores nulos, justificando la razón de acuerdo a la variable

```python
#Primer método de sustitución de valores nulos
#Sustituir valores nulos con promedio o media
#Usamos el valor promedio de la ciudad para darle un número de baños lógico y equilibrado a las viviendas sin datos.
data2["bathrooms"] = data2["bathrooms"].fillna(round(data2["bathrooms"].mean(), 1))
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Segundo método de sustitución de valores nulos
#Sustituir valores nulos con mediana
#Usamos la mediana dado que evita que los valores extremadamente altos o bajos distorsionen la información.
data2["bedrooms"]=data2["bedrooms"].fillna(round(data2["bedrooms"].median(),1))
data2["hosts_time_as_host_years"]=data2["hosts_time_as_host_years"].fillna(round(data2["hosts_time_as_host_years"].median(),1))
data2["minimum_nights"]=data2["minimum_nights"].fillna(round(data2["minimum_nights"].median(),1))
data2["maximum_nights"]=data2["maximum_nights"].fillna(round(data2["maximum_nights"].median(),1))
data2["review_scores_rating"]=data2["review_scores_rating"].fillna(round(data2["review_scores_rating"].median(),1))
data2["review_scores_location"]=data2["review_scores_location"].fillna(round(data2["review_scores_location"].median(),1))
data2["review_scores_value"]=data2["review_scores_value"].fillna(round(data2["review_scores_value"].median(),1))

```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Tercer método de sustitución de valores nulos
#Sustituir valores nulos por un valor numérico en  concreto 
#Usamos constantes lógicas de acuerda a la variable
data2["host_since"]= data2["host_since"].fillna(2026) 
data2["host_total_listings_count"] = data2["host_total_listings_count"].fillna(0)
data2["reviews_per_month"] = data2["reviews_per_month"].fillna(0)
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Cuarto método de sustitución de valores nulos
#Sustituir valores nulos por un string en  concreto 
#A las columnas de texto vacías les pusimos una etiqueta clara para saber que la información 
#no venía en la base de datos y no dejar la celda en blanco.

data2["neighbourhood"] =data2["neighbourhood"].fillna("Sin Información") 
data2["neighbourhood_group_cleansed"] =data2["neighbourhood_group_cleansed"].fillna("Sin Información") 
data2["host_location"] =data2["host_location"].fillna("Sin Información") 
data2["host_response_rate"] =data2["host_response_rate"].fillna("Sin Información") 
data2["host_acceptance_rate"] =data2["host_acceptance_rate"].fillna("Sin Información") 
data2["bathrooms_text"] =data2["bathrooms_text"].fillna("Sin Información") 
data2["instant_bookable"] =data2["instant_bookable"].fillna("Sin Información") 
data2["license"] =data2["license"].fillna("Sin Información") 
data2["has_availability"] =data2["has_availability"].fillna("Sin Información") 
data2["host_response_time"] =data2["host_response_time"].fillna("Sin Información") 
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Quinto método de sustitución de valores nulos
#Sustituir valores nulos por valores no nulos hacia adelante  "forward fill" ("ffill")
#Filtro por columnas
data2["beds"] =data2["beds"].ffill()
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sexto método de sustitución de valores nulos
#Sustituir valores nulos por valores no nulos hacia atrás backward fill" ("bfill")
#Filtro por columnas
#Rellenamos los precios faltantes tomando el costo de la propiedad siguiente como referencia de mercado.

data2["price"] =data2["price"].bfill() 
data2["price_quote_price_per_night"] =data2["price_quote_price_per_night"].bfill() 
data2["price_quote_total_price"] =data2["price_quote_total_price"].bfill() 
data2["estimated_revenue_l365d"] =data2["estimated_revenue_l365d"].bfill() 
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
data2.to_csv("OSLO_50_COLUMNAS_LIMPIO.csv", index=False)
```
